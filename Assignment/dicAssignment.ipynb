```python
# 2. Explain the concept of dictionary comprehension in Python.

"""Ans>>> 

instead of writing multiple line of the code user can create dictionary within single line using for loop concept


That help us to save time and complex coding 


 """
#Example >>>

numbers = [1, 2, 3, 4, 5]

result = {}

for num in numbers:
    result[num] = num**2

print(result)
print (result.values())
print(result.items())
print(result.keys())
print(result.keys())
```

```python
# 3. Explain the difference between the dict.get(key) method and accessing dict[key] directly.
"""
Ans -->>  dict.get(key) and dict[key] are use for Access the key's Value of the dictionary 
          The major difference of the .get() and dict[key] is the 
          -- .get() if the key has not available in the dictionary it does not return error 
              [key] if key does not present in the dictionary the it will return error 


    Example >>>>
"""


number = {2,3,4,5,6}

result = {}

for num in number:

    result[num] = num**3

print("This is dictionary",result)

print(result[6])
print(result.get(6))
print("This is also for examole dict.get()",result.get(8))

print("This is dict[key] for example purpose",result[8])
```

```python
# 5. How can you update dictionary values in Python? Provide examples.

""""There Are Two way's to update the dictionary 

1) ->> The first way we can use "=" operator we can update key's value 
2) ->> The second way we can use update method but while using update method we need to pass two argument as a tuple key and vlaue pair if key is not present in the dictionary then it will create new key value in the dictionary 


"""


number = {1,2,3,4}
result = {}
for num in number:
    result[num] = num**2

print(result)    

result[2] = 5
print(result)
result.update([(6,36)])
print("updated",result)
```

```python
# 20. Count the number of vowels in a given string and store the counts in a dictionary.

string = "Hello World python"
s = "aeiou"
result = {}

for char in s:
    result [char] = string.count(char)
print(result)    
```

```python
# 21. Create a nested dictionary where the keys are letters from 'A' to 'E' and the values are
# dictionaries mapping numbers from 1 to 5 to their squares.

letters = ["A", "B", "C", "D", "E"]

result = {
    l: {num: num**2 for num in range(1, 6)}
    for l in letters
}

print(result)
```
